```python
import sys
sys.path.append("/Users/gccasini/GitHub/diplomatrixbr-eval")
from scripts.utils import get_mean
import pandas as pd


df_eval_3r = get_mean("/Users/gccasini/GitHub/diplomatrixbr-eval/prompt_testing/sheets/sabia_2022_p7-12_3r.csv", 3)
df_eval_10r = get_mean("/Users/gccasini/GitHub/diplomatrixbr-eval/prompt_testing/sheets/sabia_2022_p7-12_10r.csv", 10)

df_std_comparison = pd.DataFrame()
df_std_comparison["nota_media_3r"] = df_eval_3r["nota_final"]
df_std_comparison["nota_media_10r"] = df_eval_10r["nota_final"]
df_std_comparison["std_3r"] = df_eval_3r["nota_final_std"]
df_std_comparison["std_10r"] = df_eval_10r["nota_final_std"]
df_std_comparison.describe()
```

```python
import plotly.graph_objects as go #plotly-6.6.0 nbformat-5.10.4
colors = {"n3": "#5A8DBE", "n10": "#E07B54"}

fig = go.Figure()

for label, col, color in [("n = 3", "std_3r", colors["n3"]),
                           ("n = 10", "std_10r", colors["n10"])]:

    # Violin
    fig.add_trace(go.Violin(
        y=df_std_comparison[col],
        name=label,
        fillcolor=color,
        opacity=0.7,
        line_color=color,
        marker=dict(color=color, size=4, opacity=0.5),
    ))

fig.update_layout(
    title=dict(
        text="Distribution of per-essay standard deviation: n=3 vs n=10",
        font=dict(size=15)
    ),
    yaxis_title="Standard deviation of scores",
    xaxis_title="Sample size condition",
    violingap=0.3,
    violinmode="overlay",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
    template="plotly_white",
    height=500,
)
```

```python
colors = {"n3": "#5A8DBE", "n10": "#E07B54"}

fig = go.Figure()

for label, col, color in [("n = 3", "std_3r", colors["n3"]),
                           ("n = 10", "std_10r", colors["n10"])]:
    fig.add_trace(go.Histogram(
        x=df_std_comparison[col],
        name=label,
        marker_color=color,
        opacity=0.6,
        nbinsx=30,
    ))

fig.update_layout(
    barmode="overlay",
    title=dict(
        text="Distribution of per-essay standard deviation: n=3 vs n=10",
        font=dict(size=15)
    ),
    xaxis_title="Standard deviation of scores",
    yaxis_title="Number of essays",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
    template="plotly_white",
    height=450,
)
```

```python
from scipy.stats import pearsonr, spearmanr
print(pearsonr(df_std_comparison["nota_media_3r"], df_std_comparison["nota_media_10r"]))
print(spearmanr(df_std_comparison["nota_media_3r"], df_std_comparison["nota_media_10r"]))
```
